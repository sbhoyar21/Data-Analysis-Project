```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(df_claims.drop('Target',axis=1),df_claims['Target'],test_size=0.20,random_state=60)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dc=DecisionTreeClassifier()
```

```python
dc.fit(x_train,y_train)
```

```python
y_pred=dc.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
print(accuracy_score(y_test,y_pred))
```

```python
from sklearn.model_selection import cross_val_score
```

```python
dc=DecisionTreeClassifier()
cross_val_score(dc,df_claims.drop('Target',axis=1),df_claims['Target'],cv=10,scoring='accuracy').mean()
```
